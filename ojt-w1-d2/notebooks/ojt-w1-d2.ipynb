```python
import torch
```

```python
x = torch.arange(24) # shape (24,)

# reshape
a = x.reshape(4,6) # shape: ([4,6])
b = x.reshape(2,3,4) # shape ([2,3,4])
c = x.reshape(2,-1) # shape([2,12]): (-1 -> tự tính size còn lại)vì x có 24 tham số, được reshape về 2 hàng -> 12 cột (24/2)

# vỉew (yêu cầu contignous)
d = x.view(6,4) # shape ([6,4])

# permute
e = torch.rand(2,3,4)
f = e.permute(2,0,1) # shape ([4,2,3]): ban đầu dữ liệu sẽ được đánh STT (0,1,2), permute sẽ thay đổi vị trí shape theo thứ tự đã đánh dấu

# View
g = e.permute(1,0,2) # non-contiguous
try:
    h = g.view(3,-1) # → RuntimeError!
except RuntimeError as err:
    print("❌ view lỗi:", err)
    # fix
    h = g.contiguous().view(3,-1)
    print("✅ Fix bằng .contiguous():", h.shape)
         
print(a.shape, b.shape, c.shape, d.shape, f.shape)
```

❌ view lỗi: view size is not compatible with input tensor's size and stride (at least one dimension spans across two contiguous subspaces). Use .reshape(...) instead.
-> Do view không thay đổi cách lưu dữ liệu của bộ nhớ mà nó sẽ thay đổi cách đọc bộ nhớ. Vì vậy khi dùng permute để thay đổi vị trí lưu dữ liệu nên view sẽ không thể đọc được
✅ Fix bằng .contiguous(): torch.Size([3, 8])
torch.Size([4, 6]) torch.Size([2, 3, 4]) torch.Size([2, 12]) torch.Size([6, 4]) torch.Size([4, 2, 3])
-> contiguous sẽ lưu 1 bản sao của tensor h vì vậy khi dùng view thì nó sẽ đọc trên cái bản sao ấy nên sẽ không lỗi

```python
import torch

def preprocess_mnist(images: torch.Tensor) -> dict:
    """
    Args:
        images: (B, 1, 28, 28) uint8 [0,255]
    Returns:
        dict với keys: 'flat', 'mean', 'std'
    """
    B = images.shape[0]

    # TODO 1: convert sang float, normalize [0,1]
    x = images.float()/255.0

    # TODO 2: flatten → (B, 784)
    flat = x.reshape(B, -1)
    # x.revew(B,-1)

    # TODO 3: mean/std theo chiều batch (dim=0)
    mean = flat.mean(dim=0) # shape: (784,)
    std = flat.std(dim=0) # shape: (784,)

    return {"flat": flat, "mean": mean, "std": std}
```

```python
# Test
dummy = torch.randint(0, 256, (64, 1, 28, 28), dtype=torch.uint8)
result = preprocess_mnist(dummy)
assert result["flat"].shape == (64, 784)
assert result["mean"].shape == (784,)
print("✅ Pass! flat:", result["flat"].shape)
```
