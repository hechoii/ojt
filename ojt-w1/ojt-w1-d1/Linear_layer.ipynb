```python
import numpy as np
from typing import Tuple
    
class LinearLayer:
    def __init__(self, int_feature: int, out_feature: int) -> None:
        self.W = np.random.rand(int_feature, out_feature)*0.01
        self.b = np.random.rand(out_feature)*0.01
        self.int_feature = int_feature
        self.out_feature = out_feature

    def forward(self, input: np.ndarray) -> np.ndarray:
        return input @ self.W + self.b # multiplication matrix

    def __repr__(self) -> str:
        return f"LinearLayer(input={self.int_feature}, out={self.out_feature})"
    

layer = LinearLayer(4,2)
x= np.random.rand(8,4)
out =layer.forward(x)
assert out.shape == (8,2)
print("Pass. Output Shape:", out.shape)
print(layer)
```

```python
class Activation:
    @staticmethod
    def relu(x: np.ndarray) -> np.ndarray:
        return np.maximum(0, x)

    @staticmethod
    def sigmoid(x: np.ndarray) -> np.ndarray:
        return 1 / (1 + np.exp(-x))
    
class MLPClassifier(Activation):
    def __init__(self, input_size: int, hidden_size: int, output_size: int) -> None:
        self.layer1 = LinearLayer(input_size, hidden_size)
        self.layer2 = LinearLayer(hidden_size, output_size)

    def predict(self, x: np.ndarray) -> np.ndarray:
        hidden_output = self.relu(self.layer1.forward(x))
        output = self.sigmoid(self.layer2.forward(hidden_output))
        return output
    
mlp = MLPClassifier(4, 5, 2)
x = np.random.rand(8, 4)
predictions = mlp.predict(x)
assert predictions.shape == (8, 2)
print("Pass. Predictions Shape:", predictions.shape)
```
